# Research Design Overview — *The Echo Emotion Language System*

---

## Research Vision  
This project explores how human emotions can be **decoded, structured, and expressed** through physiological signals.  
It seeks to transform affective computing from mere “recognition” into a **language-building process** — enabling machines not just to detect emotion, but to *speak it back* in a structured and meaningful way.  

The system’s long-term goal is to build a bridge between **neuroscience, computation, and expression** — a world where emotional states become data, and data becomes empathy.  

---

| Stage | Focus | Output | Innovation Layer |

## The Emotional Lexicon 

The **Emotional Lexicon** is a self-created dictionary that translates quantitative physiological states into symbolic, interpretable emotional units — *“emotion tokens.”*  

Each token encodes:  

| Dimension | Example | Source |
|------------|----------|--------|
| **Physiological** | HRV ratio, skin conductance | From HRV feature extraction |
| **Affective** | calm, tense, amused, fatigued | From WESAD labels + clustering |
| **Semantic** | `⟦E_Calm_01⟧`, `⟦E_Stress_07⟧` | Generated symbols for modeling |
| **Linguistic / Expressive** | mapped to words, tones, or visuals | Linked to lexicon grammar rules |

This dictionary acts as the “vocabulary” of your emotional language —  
a system where each physiological pattern corresponds to a semantic unit,  
and these units can be composed, like words, to describe nuanced emotional landscapes.  

In the final stages, this lexicon will be used to:
- Train interpretable ML models (emotion grammar learning)  
- Generate multimodal outputs (sound, visualization, or linguistic equivalents)  
- Provide a **structured bridge** between emotional experience and computational understanding

## The Musical Layer (Emotion-to-Music Translation)

This layer transforms the **emotional timeline and lexicon** into **musical compositions**.  
Using mappings between physiological features and musical dimensions, Echo creates sound that mirrors emotional dynamics.

| Emotion Dimension | Musical Mapping | Example |
|--------------------|------------------|----------|
| **Arousal / HRV variance** | Tempo | Higher arousal → faster tempo |
| **Valence / Emotional polarity** | Key signature | Positive → major; negative → minor |
| **Amplitude / Signal intensity** | Dynamics | Strong signal → louder dynamics |
| **Transition speed** | Rhythm complexity | Sudden change → syncopated rhythm |

The result is a **data-driven composition system** —  
where each subject’s emotional journey can be *heard* as a symphony of heartbeats and feelings.  

Potential technical outputs include:
- `emotion_music_map.yaml` — mapping rules between physiological features and musical parameters  
- `generate_music.py` — Python script generating MIDI via libraries such as `music21` or `mido`  
- `emotion_tracks/` — folder containing generated `.mid` or `.wav` compositions  
- `emotion_music_demo.ipynb` — notebook visualizing HRV-to-music transformation
